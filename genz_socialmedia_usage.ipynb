```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("genz_socialmedia_usage.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# Average daily usage hours
avg_daily_usage_hrs = df["daily_usage_hours"].mean()
print(avg_daily_usage_hrs)
```

```python
# Most popular platform
popular_platform = df['primary_platform'].value_counts()
print(popular_platform)
```

```python
# High addiction users
high_addiction_users = df[df["addiction_level"] == "High"]
print(high_addiction_users)
```

```python
# Country wise usage hours
country_wise_usage_hours = df.groupby("country")["daily_usage_hours"].mean().sort_values(ascending=False)
print(country_wise_usage_hours)
```

```python
# platform popularity
popular_platform.plot(kind="bar")
plt.xlabel("platform")
plt.ylabel("Number of users")
plt.title("Platform popularity")
```

```python
# Addiction levels
addiction_count = df["addiction_level"].value_counts()
plt.pie(addiction_count, labels=addiction_count.index, autopct='%1.1f%%')
plt.title("Addiction level")
```

```python
# usage vs mental health
plt.scatter(df["daily_usage_hours"], df["mental_health_score"])
plt.xlabel("daily usage hours")
plt.ylabel("mental health score")
plt.title("Usage vs Mental health score")            
```

```python
# Country wise users
country_count = df["country"].value_counts()
country_count.plot(kind="bar")
plt.xlabel("Country")
plt.ylabel("Number of users")
plt.title("Country wise users")
```

```python
# Age group and Addiction 
addiction_of_age_group = pd.crosstab(df["Age_group"], df["addiction_level"])
addiction_of_age_group.plot(kind="bar", stacked=True)
plt.xlabel("Age Group")
plt.ylabel("Number of users")
plt.title("Age Group and Addiction")
```

```python
# Mental health category
mental_health = df["Mental_health_category"].value_counts()

plt.pie(mental_health,
        labels=mental_health.index,
        autopct='%1.1f%%',
        wedgeprops=dict(width=0.4))
plt.title("Mental Health Categories")
```
